### Injecting Nemotron-3-Embed-1B-BF16 as language model of Ministral-3-3B-Instruct-2512-BF16 and reducing dims of multi_modal_projector

```python
from transformers import AutoModel
```

```python
BASE_MODEL_PATH = "/lustre/fsw/portfolios/datascience/users/gmoreira/models/Ministral-3-3B-Instruct-2512-BF16"
LLM_MODEL_TO_INJECT = "nvidia/Nemotron-3-Embed-1B-BF16"

OUTPUT_PATH = "/lustre/fsw/portfolios/datascience/users/gmoreira/models/Ministral-3-1B-Instruct-2512-BF16_llm_replaced_nemotron_3_embed_1B"
```

```python
llm_model = AutoModel.from_pretrained(LLM_MODEL_TO_INJECT, trust_remote_code=True)
llm_model
```

```python
base_model = AutoModel.from_pretrained(BASE_MODEL_PATH)
base_model
```

```python
base_model.language_model
```

```python
base_model.language_model = llm_model
```

```python
print(base_model)
```

```python
base_model.multi_modal_projector
```

```python
import torch
from torch import nn
```

```python
new_linear_1 = nn.Linear(
    1024,
    2048,
    bias=False,
    dtype=base_model.multi_modal_projector.linear_2.weight.dtype,
    device=base_model.multi_modal_projector.linear_2.weight.device,
)
new_linear_2 = nn.Linear(
    2048,
    2048,
    bias=False,
    dtype=base_model.multi_modal_projector.linear_2.weight.dtype,
    device=base_model.multi_modal_projector.linear_2.weight.device,
)
```

```python
with torch.no_grad():
    new_linear_1.weight.copy_(base_model.multi_modal_projector.linear_1.weight[:2048])
    new_linear_2.weight.copy_(base_model.multi_modal_projector.linear_2.weight[:2048, :2048])
```

```python
base_model.multi_modal_projector.linear_1 = new_linear_1
base_model.multi_modal_projector.linear_2 = new_linear_2
```

```python
base_model.multi_modal_projector.linear_1.weight.shape
```

```python
base_model.multi_modal_projector.linear_2.weight.shape
```

```python
base_model.config.text_config
```

```python
llm_model.config
```

```python
base_model.config.text_config = llm_model.config
```

```python
base_model.save_pretrained(OUTPUT_PATH)
```

```python
loaded_model = AutoModel.from_pretrained(OUTPUT_PATH)
```

```python
loaded_model
```
